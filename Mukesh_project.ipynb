Checking Missing Values

```python
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values[missing_values > 0])
```

Handling Missing Values

```python
columns_to_fill = [
    'Rainfall_mm',
    'Soil_Moisture_pct',
    'Yield_Tonnes_Ha'
]

for column in columns_to_fill:
    df[column] = df.groupby('Season')[column].transform(
        lambda x: x.fillna(x.median())
    )

print("Missing values handled successfully!")
```

```python
print("Missing Values After Treatment:")
print(df.isnull().sum())
```

One-Way ANOVA Test for Seasonal Yield

```python
from scipy.stats import f_oneway

# Create yield groups for each season
kharif_yield = df[df['Season'] == 'Kharif']['Yield_Tonnes_Ha']
rabi_yield = df[df['Season'] == 'Rabi']['Yield_Tonnes_Ha']
zaid_yield = df[df['Season'] == 'Zaid']['Yield_Tonnes_Ha']

# Perform One-Way ANOVA
f_statistic, p_value = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("F-Statistic:", round(f_statistic, 4))
print("P-Value:", p_value)
```

ANOVA Hypothesis Test Conclusion

```python
if p_value < 0.05:
    print("Reject the Null Hypothesis (H₀)")
    print("Conclusion: There is a statistically significant difference in crop yield between seasons.")
else:
    print("Fail to Reject the Null Hypothesis (H₀)")
    print("Conclusion: There is no statistically significant difference in crop yield between seasons.")
```
